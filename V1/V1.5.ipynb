```python
from common import *
%load_ext autoreload
%autoreload 2
sizes = [16, 4, 512]
import matplotlib
matplotlib.rcParams['figure.figsize'] = [10, 8]
```

```python
learning_rates = (1e-2, 1e-3, 1e-4)

dim = 3000
x_train, y_train, x_test, y_test = load_dataset(dim)
models = []
for s in sizes:
    model = init_model(dim, s, 0.5, keras.regularizers.l2())
    train(model, x_train, y_train, learning_rates)
    models.append(model)
```

```python
dim = 5000

x_train, y_train, x_test, y_test = load_dataset(dim)
models = []
for s in sizes:
    model = init_model(dim, s, 0.0)
    train(model, x_train, y_train, learning_rates)
    models.append(model)
    model = init_model(dim, s, 0.0, keras.regularizers.l2())
    train(model, x_train, y_train, learning_rates)
    models.append(model)
    model = init_model(dim, s, 0.5, None)
    train(model, x_train, y_train, learning_rates)
    models.append(model)
```

```python
for s in sizes:
    plot_results(f'model_3000_{s}', f'model_3000_{s}_l2', f'model_3000_{s}_do0.5', f'model_3000_{s}_l2_do0.5', learning_rates=learning_rates)
```

```python
for s in sizes:
    plot_results(f'model_3000_{s}', f'model_5000_{s}', learning_rates=learning_rates)
    plot_results(f'model_3000_{s}_l2', f'model_5000_{s}_l2', learning_rates=learning_rates)
    plot_results(f'model_3000_{s}_do0.5', f'model_5000_{s}_do0.5', learning_rates=learning_rates)
```
